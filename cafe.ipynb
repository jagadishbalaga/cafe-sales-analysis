```python
# import pandas
import pandas as pd
```

```python
# load the data
df = pd.read_csv("cafe_sales_practice.csv")
```

```python
df
```

```python
df.columns = df.columns.str.lower().str.title()
```

```python
df['Category'] = df['Category'].str.title()
```

```python
df['Category'].unique()
```

```python
category = {
    'Beverage':'Beverages'
}
```

```python
df['Category'] = df['Category'].replace(category)
```

```python
df['Category'].unique()
```

```python
# count the nulls
df.isnull().sum()
```

```python
df['CustomerName']
```

```python
df['CustomerName'] = df['CustomerName'].fillna('Unknown')
```

```python
df['CustomerName']
```

```python
unit_price = df['UnitPrice'].mean().round(0)
```

```python
unit_price
```

```python
df['UnitPrice'] = df['UnitPrice'].fillna(unit_price)
```

```python
df.info()
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
```

```python
df.isnull().sum()
```

```python
quantity = df['Quantity'].median()
```

```python
# fill quantity 
df['Quantity'] = df['Quantity'].fillna(quantity)
```

```python
total = df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
```

```python
df['TotalPrice'] = df['TotalPrice'].fillna(total)
```

```python
df['PaymentMethod'] = df['PaymentMethod'].fillna('Unknown')
```

```python
df['PaymentMethod'].isnull().sum()
```

```python
rating = df['Rating'].median()
```

```python
df['Rating'] = df['Rating'].fillna(rating)
```

```python
df.isnull().sum()
```

```python
df
```

```python
df['Item'].value_counts()
```

```python
# remove duplicate rows from your dataset.
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# groupby() to find total revenue for Category, City, and Month.
# Revenue by Category
revenue_by_category = df.groupby('Category')['TotalPrice'].sum().sort_values(ascending=False)
print(revenue_by_category)
```

```python
# Revenue by City
revenue_by_city = df.groupby('City')['TotalPrice'].sum().sort_values(ascending=False)
print(revenue_by_city)
```

```python
# Revenue by Month
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

revenue_by_month = df.groupby(df['OrderDate'].dt.month)['TotalPrice'].sum()
print(revenue_by_month)
```
