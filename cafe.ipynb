```python
# import pandas
import pandas as pd
```

```python
# load the data
df = pd.read_csv("cafe_sales_practice.csv")
```

```python
df
```

```python
# count the nulls
df.isnull().sum()
```

```python
# drop rows where custmoername is missing
df = df.dropna(subset=['CustomerName'])
```

```python
# fill quantity with 1
df['Quantity'] = df['Quantity'].fillna(1)
```

```python
# Fill Rating with median
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
```

```python
# Fill PaymentMethod with Unknown
df['PaymentMethod'] = df['PaymentMethod'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['UnitPrice'] = df.groupby('Item')['UnitPrice'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df['TotalPrice'] = df['TotalPrice'].fillna(
    df['Quantity'] * df['UnitPrice']
)
```

```python
df.isnull().sum()
```

```python
df['City'] = df['City'].fillna(df['City'].mode()[0])
```

```python
# Clean the Category column so all category names have the same format
df['Category'] = df['Category'].str.strip().str.title()
```

```python
df['Category'].value_counts()
```

```python
# It means remove unnecessary spaces from the beginning and end of the Item names.
df['Item'] = df['Item'].str.strip()
```

```python
df['Item'].value_counts()
```

```python
# remove duplicate rows from your dataset.
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# TotalPrice is missing, calculate it again using Quantity × UnitPrice.
df['TotalPrice'] = df['TotalPrice'].fillna(
    df['Quantity'] * df['UnitPrice']
)
```

```python
df['TotalPrice'].isnull().sum()
```

```python
# groupby() to find total revenue for Category, City, and Month.
# Revenue by Category
revenue_by_category = df.groupby('Category')['TotalPrice'].sum().sort_values(ascending=False)
print(revenue_by_category)
```

```python
# Revenue by City
revenue_by_city = df.groupby('City')['TotalPrice'].sum().sort_values(ascending=False)
print(revenue_by_city)
```

```python
# Revenue by Month
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

revenue_by_month = df.groupby(df['OrderDate'].dt.month)['TotalPrice'].sum()
print(revenue_by_month)
```
